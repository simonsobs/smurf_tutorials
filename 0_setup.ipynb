## Making good RF connections and cable management

All RF connections should either be made finger tight, or with an SMA wrench with preset torque.  NEVER TIGHTEN RF CONNECTIONS WITH A regular wrench.  Remember that these are very expensive parts.  Here's a <a href="https://www.thorlabs.com/newgrouppage9.cfm?objectgroup_ID=10543">cheap one I like from Thorlabs</a> if your lab doesn't already have some (the 5 in-lb one is for making connections when you're testing, while the 8 in-lb one is for making final connections once you're happy with your RF chain).  Reduce the temptation to take shortcuts by purchasing more than you need!


<img style="transform: rotate(90deg); width:1000px" src="setup_pics/rf_torque_wrench.JPG" alt="RF Torque Wrench" title="RF Torque Wrench" />

A "feature" of the AMCs is that they have very long barrel connectors for the frontpanel RF connections - e.g.<br>
<img style="transform: rotate(0deg); width:1000px" src="setup_pics/long_barrel_rf_connectors.JPG" alt="SMuRF AMC Long Barrel AMC RF Connectors" title="SMuRF AMC Long Barrel AMC RF Connectors" />

On at least one occasion, Shawn has snapped one of these off when tightening an SMA cable to it.  It's also easy to put a bunch of torque on the connector (big lever arm) if you bump a straight SMA cable connected to one of them, or if you don't support the connector as you fasten or remove an SMA connection.  A trick that helps is to put a right angle SMA adapter on the connector to shorten the lever arm and redirect the force.  It also keeps the cables close to the front of the crate so it's less likely you'll bump them accidentally.  E.g.:

<img style="transform: rotate(0deg); width:1000px" src="setup_pics/right_angle_rf_connections.JPG" alt="Right Angle RF Connections" title="Right Angle RF Connections" />

Putting some thought into your cable management is critical for protecting the SMuRF boards from damage.  Here's a picture of how the SMuRF blade was connected up in the Stanford lab (at least as of the morning of 1/30/19):

<img style="transform: rotate(0deg); width:1000px" src="setup_pics/cable_management.JPG" alt="Cable Management" title="Cable Management" />

This isn't the most perfect cable management, but illustrates a couple of things you can do to keep things sane and protect the electronics (which, I cannot stress enough, are very expensive):
<ol>
    <li>All cables are either ziptied to the crate before going off to the cryostat, or connected to something (like the blue directional coupler zip tied to the crate handle on the right with the yellow and green ties) that is firmly attached to the crate or whatever the crate is sitting on/in.  This ensures that if one of the cables get yanked on the cryostat side, the SMuRF blade is protected.</li>
    <li>The connections directly to the SMuRF are made with short, easily replaced semirigid cables.  Use cables or connector savers so that you minimize the amount of torque and wear and tear on the AMC connectors.  Short cables like this are nice because if you use rigid connector savers you still can exert quite a bit of force on the AMC connectors when making/breaking connections.  If you use short flex cables like this, be sure to check them often.</li>
    <li>Right angle connector savers on the AMC connectors help reduce the lever arm of the RF cables, and keeps the connectors are cables relatively flush with the front of the crate, keeping them out of the way.</li>
    <li>Always screw in the carriers and RTMs into the crate.  This keeps them from moving around when you're making/breaking connections to them.  Don't tighten more than finger tight.</li>
    <li>Never cantilever something off of an RF connector like we did (the blue box with the fiber coming out of it) in this picture.  That connector broke (duh).
</ol>

# Initializing the system (jackhammer)

In sodetlib there is a set of utility functions called `jackhammer`. To start the system from scrath run `jackhammer hammer` on the smurf server (in person, over ssh, or vnc **but not in jupyter**) this does:

1. Kills any existing smurf or ocs dockers
2. In a utility docker:
 - Sets fan speeds based on sys_config file
 - Deactivates and reactivates slots
 - waits for slots to come back online
3. Restarts all smurf and ocs dockers
4. Runs "setup" on pysmurf for each slot
5. Dumps you into a ipython session for the first smurf slot specified in the sys_config file

Note that some of these steps take a while to complete, and happen within dockers, so they will not be printing to stdout. If you'd like to view what's going on inside the docker containers, you can run `jackhammer logs` to view all docker logs, or `jackhammer logs <service_name>` to view logs for a specific service. For instance: `jackhammer logs smurf-streamer-s2` will show logs inside the streamer docker container. You can also use `docker logs` or `docker-compose logs` to view them in the same way if you are familiar with those commands.

if you already have a all pysmurf dockers running and just want to instantiate another pysmurf instance you can run `jackhammer pysmurf`

if you want to pull up the rogue gui you can use `jackhammer gui`

if you just want to restart the dockers but not the hardware you can use `jackhammer soft_reset`

Once you run `jackhammer hammer` to start your dockers on the smurf server you can instantiate a pysmurf control object with the following cell (refer to the sodetlib tutorial notebook for more detail on this step).

````python
from sodetlib.det_config import DetConfig
cfg = DetConfig()
# configuration is the topic of the next notebook
cfg.load_config_files(slot=4)  # for a carrier in slot 4
S = cfg.get_smurf_control()
````

## Cold and warm RF amplifiers

You can check the bias voltages and currents for each amplifier with the following command:

````python
S.get_amplifier_biases()
````

This reports the 50K and 4K amplifier gate voltages and measured drain currents.  According to their datasheets, the 4K LNF HEMT we're using should have a drain current of 4mA, while the 50K LNA should have a drain current of 21mA. Do reach these targets, you can adjust the voltages through pysmurf using this function;

**The appropriate currents will depend on the amplifiers you are using.**

````python
# aim for 4mA
S.set_amp_drain_voltage('hemt1',0.5)
S.set_amp_gate_voltage('hemt1',0.25,override=True)

# Bias RFC1 50K LNA -- aim for 21mA
S.set_amp_drain_voltage('50k1',5.0)
S.set_amp_gate_voltage('50k1',0.0)
````

````python
S.get_amplifier_biases()
````

Keep in mind that the drain currents do vary depending on how much power is circulating through the amplifier and its temperature.

## Locking to external frequency reference

**TODO: This is section is out of date, but still contains useful info.**

The SMuRF system can optionally run off an external frequency reference.  In testing, we've discovered that this can substantially improve RF performance (still under investigation).  For this purpose we recommend the <a href="https://www.thinksrs.com/products/cg635.html">Stanford Research Systems CG635 2.05 GHz Synthesized Clock Generator</a>.  This is the same reference generator that's currently being used on BKuMUX at the South Pole, where it is working well.  Once you plug the reference into SMuRF (either on the front of the AMC or through a timing AMC and carrier, as in our campus demo setup), it's important to make sure that you're actually locked to the reference.  A simple check is to generate some tones with the SMuRF with feedback turned off, look closely at the tone on a signal analyzer, and verify that when you switch off the reference, the tone frequency changes (indicating you've lost the reference).  For instance, if you are locked and tracking some tones with the flux ramp on, you can turn off the flux ramp and feedback like this:

````python
S.set_feedback_enable(2,0)
S.flux_ramp_off()
````

You can check that this worked with the corresponding `get` command:

````python
S.get_feedback_enable(2)
````

Here's an example of this check (looking at tones with flux ramp and feedback off, after going through the cryo and resonators):<br>
<img style="transform: rotate(0deg); width:600px" src="setup_pics/reference_unlocked_vs_locked.PNG" alt="Fixed SMuRF Tones Locked to Reference VS Unlocked" title="Fixed SMuRF Tones Locked to Reference VS Unlocked" />

The blue is with the frequencies set to the minimum of each tones' resonator, while the yellow is what happens when you switch the external reference (which the signal analyzer is also locked to) off.  The tones shift off resonance, increasing the total power transmitted to the signal analyzer.  If you turn off your external reference and the tones don't move you're not locked to it!
